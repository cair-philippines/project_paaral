# 0.9.1 Provincial Network Builder (SciPy Optimized)

This notebook demonstrates the use of **Module 12.1: ProvincialNetworkBuilderSciPy** to create road network graphs and distance matrices using **scipy.sparse.csgraph** for ~10x speedup.

**Key Improvements over 0.9:**
- ⚡ **~10x faster** distance matrix computation using scipy
- Uses scipy.sparse.csgraph.dijkstra() instead of NetworkX loops
- No multiprocessing needed (scipy is already optimized in C/Cython)
- Lower memory footprint with sparse matrix representation

**Performance:**
- NetworkX version (0.9): Minutes for 1000 schools
- SciPy version (0.9.1): Seconds for 1000 schools

**Example Province:** Bulacan (PH03014)

**Outputs:**
1. Distance matrix (CSV)
2. Distance graph (GraphML)
3. Beneficiary graph (GraphML)
4. Summary statistics (JSON)
5. Visualizations

```python
%%time
# Cell 42.1 (Code) - Load and prepare unified flows
print("Loading unified Grade 7 flows for graph visualization...")

# Import unified flow builder
from modules.unified_gr7_flow_builder import UnifiedGr7FlowBuilder

# Initialize builder with paths
unified_builder = UnifiedGr7FlowBuilder(
    beneficiary_parquet_path='data/processed/esc_beneficiaries.parquet',
    gr7_enrollees_path='output/gr7_enrollees_sy2023_2024.csv',
    public_nodes_path='output/public_nodes_valid.gpkg',
    private_nodes_path='output/private_nodes_valid.gpkg',
    verbose=True
)

# Build unified table
print("\nBuilding unified Grade 7 table...")
unified_df = unified_builder.build_unified_table(school_year='SY 2023-2024')

# Aggregate for graph creation
print("\nAggregating flows for graph creation...")
aggregated_flows = unified_builder.aggregate_for_graph()

print(f"\n✓ Unified flows loaded and aggregated")
print(f"  Student-level records: {len(unified_df):,}")
print(f"  Aggregated school pairs: {len(aggregated_flows):,}")
print(f"  Total beneficiaries: {aggregated_flows['beneficiary_count'].sum():,}")
print(f"  Total non-beneficiaries: {aggregated_flows['non_beneficiary_count'].sum():,}")
```

```python
%%time
# Cell 42.2 (Code) - Filter to province and create flow geometries
print(f"Filtering unified flows to {PROVINCE_NAME.title()} province ({PROVINCE_CODE})...")

# Get school IDs in current province
bulacan_school_ids_set = set(public_bulacan['school_id']) | set(private_bulacan['school_id'])

# Filter to province flows (where origin OR destination is in province)
province_flows = aggregated_flows[
    aggregated_flows['school_id_origin'].isin(bulacan_school_ids_set) |
    aggregated_flows['school_id_destination'].isin(bulacan_school_ids_set)
].copy()

# Further filter to in-province flows only (both origin AND destination in province)
in_province_flows = province_flows[
    province_flows['school_id_origin'].isin(bulacan_school_ids_set) &
    province_flows['school_id_destination'].isin(bulacan_school_ids_set)
].copy()

print(f"\n✓ Province filtering:")
print(f"  Total flows touching province: {len(province_flows):,}")
print(f"  In-province flows only: {len(in_province_flows):,}")
print(f"  Cross-province flows: {len(province_flows) - len(in_province_flows):,}")
print(f"  Total students in-province: {in_province_flows['total_student_count'].sum():,}")
print(f"    Beneficiaries: {in_province_flows['beneficiary_count'].sum():,}")
print(f"    Non-beneficiaries: {in_province_flows['non_beneficiary_count'].sum():,}")

# Combine all schools for coordinate lookup
all_schools = pd.concat([public_bulacan, private_bulacan], ignore_index=True)

# Create flow geometries
print(f"\nCreating flow geometries...")

def create_unified_flow_gdf(flow_df, all_schools_gdf):
    """Create GeoDataFrame with flow geometries"""
    flow_geoms = []
    skipped = 0

    for idx, row in flow_df.iterrows():
        origin_id = row['school_id_origin']
        dest_id = row['school_id_destination']

        # Get coordinates from schools
        origin_row = all_schools_gdf[all_schools_gdf['school_id'] == origin_id]
        dest_row = all_schools_gdf[all_schools_gdf['school_id'] == dest_id]

        if len(origin_row) > 0 and len(dest_row) > 0:
            origin_geom = origin_row.iloc[0].geometry
            dest_geom = dest_row.iloc[0].geometry

            # Create LineString
            line = LineString([origin_geom.coords[0], dest_geom.coords[0]])

            flow_geoms.append({
                'origin_id': origin_id,
                'dest_id': dest_id,
                'beneficiary_count': row['beneficiary_count'],
                'non_beneficiary_count': row['non_beneficiary_count'],
                'total_student_count': row['total_student_count'],
                'geometry': line
            })
        else:
            skipped += 1

    gdf = gpd.GeoDataFrame(flow_geoms, crs='EPSG:4326')
    print(f"  Created {len(gdf):,} flow lines (skipped {skipped} due to missing coordinates)")

    return gdf

# Create flow GeoDataFrame
flows_gdf = create_unified_flow_gdf(in_province_flows, all_schools)

print(f"\n✓ Flow geometries created: {len(flows_gdf):,} flows")
```

```python
# %%time
# Cell 42.3 (Code) - Plot unified flow network with purple (beneficiary) and light gray (non-beneficiary)
# Visualization 3: Unified Flow Network (Beneficiary = Purple, Non-Beneficiary = Light Gray)
print("Creating unified flow network visualization...")

# Separate beneficiary and non-beneficiary flows
# Flows with beneficiaries only (purple)
beneficiary_only_flows = flows_gdf[
    (flows_gdf['beneficiary_count'] > 0) &
    (flows_gdf['non_beneficiary_count'] == 0)
].copy()

# Flows with non-beneficiaries only (light gray)
non_beneficiary_only_flows = flows_gdf[
    (flows_gdf['beneficiary_count'] == 0) &
    (flows_gdf['non_beneficiary_count'] > 0)
].copy()

# Mixed flows (both beneficiaries and non-beneficiaries) - split into two separate edges
# We'll plot beneficiary portion as purple and non-beneficiary portion as gray
mixed_flows = flows_gdf[
    (flows_gdf['beneficiary_count'] > 0) &
    (flows_gdf['non_beneficiary_count'] > 0)
].copy()

print(f"\n✓ Flow categorization:")
print(f"  Beneficiary-only flows: {len(beneficiary_only_flows):,} ({beneficiary_only_flows['beneficiary_count'].sum():,} students)")
print(f"  Non-beneficiary-only flows: {len(non_beneficiary_only_flows):,} ({non_beneficiary_only_flows['non_beneficiary_count'].sum():,} students)")
print(f"  Mixed flows: {len(mixed_flows):,} ({mixed_flows['beneficiary_count'].sum():,} benef + {mixed_flows['non_beneficiary_count'].sum():,} non-benef)")

# Normalize line widths (log scale for better visualization)
min_width = 0.5
max_width = 3.5

# Combine all counts for consistent normalization
all_benef_counts = pd.concat([
    beneficiary_only_flows['beneficiary_count'] if len(beneficiary_only_flows) > 0 else pd.Series([]),
    mixed_flows['beneficiary_count'] if len(mixed_flows) > 0 else pd.Series([])
])

all_non_benef_counts = pd.concat([
    non_beneficiary_only_flows['non_beneficiary_count'] if len(non_beneficiary_only_flows) > 0 else pd.Series([]),
    mixed_flows['non_beneficiary_count'] if len(mixed_flows) > 0 else pd.Series([])
])

# Normalize beneficiary flows
if len(all_benef_counts) > 0:
    log_benef = np.log1p(all_benef_counts)
    benef_min = log_benef.min()
    benef_max = log_benef.max()

    if len(beneficiary_only_flows) > 0:
        beneficiary_only_flows['line_width'] = np.log1p(beneficiary_only_flows['beneficiary_count'])
        beneficiary_only_flows['line_width'] = min_width + (beneficiary_only_flows['line_width'] - benef_min) / (benef_max - benef_min) * (max_width - min_width)

    if len(mixed_flows) > 0:
        mixed_flows['benef_line_width'] = np.log1p(mixed_flows['beneficiary_count'])
        mixed_flows['benef_line_width'] = min_width + (mixed_flows['benef_line_width'] - benef_min) / (benef_max - benef_min) * (max_width - min_width)

# Normalize non-beneficiary flows
if len(all_non_benef_counts) > 0:
    log_non_benef = np.log1p(all_non_benef_counts)
    non_benef_min = log_non_benef.min()
    non_benef_max = log_non_benef.max()

    if len(non_beneficiary_only_flows) > 0:
        non_beneficiary_only_flows['line_width'] = np.log1p(non_beneficiary_only_flows['non_beneficiary_count'])
        non_beneficiary_only_flows['line_width'] = min_width + (non_beneficiary_only_flows['line_width'] - non_benef_min) / (non_benef_max - non_benef_min) * (max_width - min_width)

    if len(mixed_flows) > 0:
        mixed_flows['non_benef_line_width'] = np.log1p(mixed_flows['non_beneficiary_count'])
        mixed_flows['non_benef_line_width'] = min_width + (mixed_flows['non_benef_line_width'] - non_benef_min) / (non_benef_max - non_benef_min) * (max_width - min_width)

print(f"\n✓ Line widths normalized")

# Create figure
fig, ax = plt.subplots(figsize=(14, 12))

# Plot province boundary (base layer)
if len(province_boundary) > 0:
    province_boundary.boundary.plot(ax=ax, color='black', linewidth=2, label='Province Boundary', zorder=1)

# Plot road network (gray, very thin - context only)
road_network_gdf.plot(ax=ax, color='gray', linewidth=0.3, alpha=0.15, label='Road Network', zorder=2)

# Plot non-beneficiary flows first (LIGHT GRAY - bottom layer)
non_benef_count = 0
if len(non_beneficiary_only_flows) > 0:
    segments_non_benef = [list(geom.coords) for geom in non_beneficiary_only_flows.geometry]
    widths_non_benef = non_beneficiary_only_flows['line_width'].values

    lc_non_benef = LineCollection(
        segments_non_benef,
        linewidths=widths_non_benef,
        colors='lightgray',
        alpha=0.6,
        zorder=3,
        label=f'Non-Beneficiary Flows ({non_beneficiary_only_flows["non_beneficiary_count"].sum():,} students)'
    )
    ax.add_collection(lc_non_benef)
    non_benef_count += len(non_beneficiary_only_flows)

# Plot non-beneficiary portion of mixed flows (LIGHT GRAY - bottom layer)
if len(mixed_flows) > 0:
    segments_mixed_non_benef = [list(geom.coords) for geom in mixed_flows.geometry]
    widths_mixed_non_benef = mixed_flows['non_benef_line_width'].values

    lc_mixed_non_benef = LineCollection(
        segments_mixed_non_benef,
        linewidths=widths_mixed_non_benef,
        colors='lightgray',
        alpha=0.4,
        zorder=3
    )
    ax.add_collection(lc_mixed_non_benef)
    non_benef_count += len(mixed_flows)

# Plot beneficiary flows on top (PURPLE - top layer)
benef_count = 0
if len(beneficiary_only_flows) > 0:
    segments_benef = [list(geom.coords) for geom in beneficiary_only_flows.geometry]
    widths_benef = beneficiary_only_flows['line_width'].values

    lc_benef = LineCollection(
        segments_benef,
        linewidths=widths_benef,
        colors='purple',
        alpha=0.7,
        zorder=4,
        label=f'Beneficiary Flows ({beneficiary_only_flows["beneficiary_count"].sum():,} students)'
    )
    ax.add_collection(lc_benef)
    benef_count += len(beneficiary_only_flows)

# Plot beneficiary portion of mixed flows on top (PURPLE - top layer)
if len(mixed_flows) > 0:
    segments_mixed_benef = [list(geom.coords) for geom in mixed_flows.geometry]
    widths_mixed_benef = mixed_flows['benef_line_width'].values

    lc_mixed_benef = LineCollection(
        segments_mixed_benef,
        linewidths=widths_mixed_benef,
        colors='purple',
        alpha=0.5,
        zorder=4
    )
    ax.add_collection(lc_mixed_benef)
    benef_count += len(mixed_flows)

# Plot schools (color-coded by sector)
public_bulacan.plot(ax=ax, color='blue', markersize=12, alpha=0.7, label='Public Schools', zorder=5)
private_bulacan.plot(ax=ax, color='red', markersize=12, alpha=0.7, label='Private Schools', zorder=5)

# Styling
title_text = f'Unified Grade 7 Flow Network - {PROVINCE_NAME.title()} Province ({PROVINCE_CODE})\n'
title_text += f'(Purple = Beneficiary, Light Gray = Non-Beneficiary | Line thickness ∝ log(students))'
ax.set_title(title_text, fontsize=16, fontweight='bold', pad=20)
ax.set_xlabel('Longitude', fontsize=12)
ax.set_ylabel('Latitude', fontsize=12)
ax.legend(loc='upper right', fontsize=10, framealpha=0.9)
ax.grid(alpha=0.2, linestyle='--')

# Auto-scale to data
ax.autoscale()

plt.tight_layout()
plt.show()

print(f"\n✓ Unified flow visualization complete")
print(f"  Province: {PROVINCE_NAME.title()} ({PROVINCE_CODE})")
print(f"  Beneficiary flow edges: {benef_count:,} (purple)")
print(f"  Non-beneficiary flow edges: {non_benef_count:,} (light gray)")
print(f"  Total flows visualized: {len(flows_gdf):,}")
print(f"  Total students: {flows_gdf['total_student_count'].sum():,}")
print(f"    Beneficiaries: {flows_gdf['beneficiary_count'].sum():,}")
print(f"    Non-beneficiaries: {flows_gdf['non_beneficiary_count'].sum():,}")
```

```python
# Classify schools efficiently (vectorized approach)
print("Classifying schools by role in beneficiary network...")

# Reuse flows_gdf from Viz 3 - calculate in/out totals
in_flow_totals = flows_gdf.groupby('dest_id')['beneficiary_count'].sum().to_dict()
out_flow_totals = flows_gdf.groupby('origin_id')['beneficiary_count'].sum().to_dict()

# Get in/out edge counts from beneficiary_graph
in_counts = dict(beneficiary_graph.in_degree())
out_counts = dict(beneficiary_graph.out_degree())

# Combine all schools and add classification columns
all_schools_viz4 = pd.concat([public_bulacan, private_bulacan], ignore_index=True)
all_schools_viz4['in_count'] = all_schools_viz4['school_id'].map(in_counts).fillna(0).astype(int)
all_schools_viz4['out_count'] = all_schools_viz4['school_id'].map(out_counts).fillna(0).astype(int)
all_schools_viz4['in_beneficiaries'] = all_schools_viz4['school_id'].map(in_flow_totals).fillna(0).astype(int)
all_schools_viz4['out_beneficiaries'] = all_schools_viz4['school_id'].map(out_flow_totals).fillna(0).astype(int)

# Check if 'offers_jhs' column exists, if not create based on available data
if 'offers_jhs' not in all_schools_viz4.columns:
    print("  ⚠ 'offers_jhs' column not found, using heuristic based on sector")
    # Assume all public schools offer JHS, check private school names/other indicators
    all_schools_viz4['offers_jhs'] = all_schools_viz4['sector'] == 'Public'
else:
    # Ensure boolean type
    all_schools_viz4['offers_jhs'] = all_schools_viz4['offers_jhs'].fillna(False).astype(bool)

# Classify into 4 categories (mutually exclusive for clear visualization)
dest_jhs_mask = (all_schools_viz4['in_count'] > 0) & (all_schools_viz4['offers_jhs'] == True)
origin_only_mask = (all_schools_viz4['out_count'] > 0) & (all_schools_viz4['in_count'] == 0)
both_mask = (all_schools_viz4['in_count'] > 0) & (all_schools_viz4['out_count'] > 0) & (~dest_jhs_mask)
no_edges_mask = (all_schools_viz4['in_count'] == 0) & (all_schools_viz4['out_count'] == 0)

dest_jhs = all_schools_viz4[dest_jhs_mask].copy()
origin_only = all_schools_viz4[origin_only_mask].copy()
both_in_out = all_schools_viz4[both_mask].copy()
no_edges_schools = all_schools_viz4[no_edges_mask].copy()

print(f"✓ School classification complete:")
print(f"  Destination JHS (diamond): {len(dest_jhs)} schools")
print(f"  Origin only (circle): {len(origin_only)} schools")
print(f"  Both in/out (square): {len(both_in_out)} schools")
print(f"  No edges (greyed circle): {len(no_edges_schools)} schools")
print(f"  Total: {len(all_schools_viz4)} schools")
```
