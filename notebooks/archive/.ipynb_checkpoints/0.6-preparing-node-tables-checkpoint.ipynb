```python
def modified_coc_transform(df_wide: pd.DataFrame, modified_coc: str) -> pd.DataFrame:
    """Turn zeros to NaNs according to a school's modified_coc."""
    mask = df_wide['modified_coc'] == modified_coc
    df_tmp = df_wide.loc[mask].copy()

    if modified_coc == 'Purely ES':
        df_tmp.loc[:, 'enrollment_g7':] = np.nan
    elif modified_coc == 'Purely JHS':
        df_tmp.loc[:, 'enrollment_k':'enrollment_ng_es'] = np.nan
        df_tmp.loc[:, 'enrollment_g11':] = np.nan
    elif modified_coc == 'JHS with SHS':
        df_tmp.loc[:, 'enrollment_k':'enrollment_ng_es'] = np.nan
    elif modified_coc == 'Purely SHS':
        df_tmp.loc[:, 'enrollment_k':'enrollment_ng_jhs'] = np.nan
    elif modified_coc == 'All Offering':
        return df_tmp
    elif modified_coc == 'ES and JHS':
        df_tmp.loc[:, 'enrollment_g11':] = np.nan

    return df_tmp
```

```python
dfs = []
for coc in pvt_enr['modified_coc'].unique():
    df_tmp = modified_coc_transform(pvt_enr, coc)
    dfs.append(df_tmp)

pivot_enr = pd.concat(dfs)
print(pivot_enr.shape)
```

```python
def nans_to_zero(row, columns: List[str] = ['seats_es','seats_jhs','seats_shs']):
    bool_vals = row[['offers_es','offers_jhs','offers_shs']].values.tolist()
    seats_vals = row[columns].values

    # Pure ES
    if bool_vals == [True, False, False]:
        seats_vals[1:] = np.nan
    # ES and JHS
    elif bool_vals == [True, True, False]:
        seats_vals[2:] = np.nan
    # K to 12
    elif bool_vals == [True, True, True]:
        pass
    # JHS only
    elif bool_vals == [False, True, False]:
        seats_vals[0] = np.nan
        seats_vals[2] = np.nan
    # SHS only
    elif bool_vals == [False, False, True]:
        seats_vals[:-1] = np.nan
    # JHS and SHS
    elif bool_vals == [False, True, True]:
        seats_vals[:-2] = np.nan

    return seats_vals
```

```python
def get_difference_school_ids(ids_list_a: List=None, ids_list_b: List=None) -> List:
    diff_ids = set(ids_list_a).difference(ids_list_b)
    return diff_ids
```

### Merge

```python
# Pseudo validation for enrollment data
pivot_enr['has_enrollment_data'] = True

tmp_enr_seats = (
    pivot_enr
    .rename(columns={'school_id_processed':'school_id'})
    .merge(
        pvt_seats,
        left_on='school_id',
        right_on='school_id',
        how='left'
    )
)

# Aggregate ES, JHS, and SHS enrollment
tmp_enr_seats['enrollment_es'] = tmp_enr_seats.loc[:, 'enrollment_k':'enrollment_g6'].sum(axis=1)
tmp_enr_seats['enrollment_jhs'] = tmp_enr_seats.loc[:, 'enrollment_g7':'enrollment_g10'].sum(axis=1)
tmp_enr_seats['enrollment_shs'] = tmp_enr_seats.loc[:, 'enrollment_g11':'enrollment_g12'].sum(axis=1)

# Validate
diff_enr_seats = get_difference_school_ids(enr_ids, seats_ids)
mask = tmp_enr_seats['school_id'].isin(diff_enr_seats)
tmp_enr_seats.loc[mask, 'has_seats_data'] = False
tmp_enr_seats.loc[~mask, 'has_seats_data'] = True
```

```python
# Add facilities data
tmp_enr_faci = tmp_enr_seats.merge(
    facilities_data,
    left_on='school_id',
    right_on='school_id',
)

# Order columns
start_cols = ['school_id','modified_coc','offers_es','offers_jhs','offers_shs']
other_cols = [col for col in tmp_enr_faci.columns if col not in start_cols]
order_cols = start_cols + other_cols
tmp_enr_faci = tmp_enr_faci[order_cols]

# Nan-ify seats data based on curricular offering
columns = ['seats_es','seats_jhs','seats_shs']
tmp_enr_faci[columns] = np.vstack(tmp_enr_faci.apply(nans_to_zero, axis=1, args=(columns,)))
# Nan-ify enrollment data based on curricular offering
columns = ['enrollment_es','enrollment_jhs','enrollment_shs']
tmp_enr_faci[columns] = np.vstack(tmp_enr_faci.apply(nans_to_zero, axis=1, args=(columns,)))

# Validate
# diff_ids = get_difference_school_ids(enr_ids, faci_ids)
cols = [col for col in facilities_data if 'classroom' in col]
mask = facilities_data[cols].isna().apply(lambda row: all(row), axis=1)
tmp_enr_faci.loc[mask, 'has_facilities_data'] = False
tmp_enr_faci.loc[~mask, 'has_facilities_data'] = True
```

### ESC

```python
pvt_esc = esc_data.pivot_table(
    index=['school_id','School Name'],
    columns='fee_type',
    values='amount',
    aggfunc='mean',
    observed=False
)

# Rename columns to be ESC
pvt_esc.columns = ['esc_average_misc_fees','esc_average_other_fees','esc_average_tuition_fees']

pvt_esc = pvt_esc.reset_index().rename(columns={'School Name':'school_name_esc'})
pvt_esc.columns.name = ''

print(pvt_esc.shape)
display(pvt_esc.head())
```

### SHSVP

```python
pvt_shsvp = shsvp_data.groupby(['school_id','School Name']).agg(
    {
        'Tuition':'mean',
        'Other':'mean',
        'Miscellaneous':'mean',
    }
)

# Rename columns to be SHSVP
col_names = ['shsvp_average_tuition_fees', 'shsvp_average_other_fees', 'shsvp_average_misc_fees']
pvt_shsvp.columns = col_names

pvt_shsvp = pvt_shsvp.reset_index().rename(columns={'School Name':'school_name_shsvp'})
pvt_shsvp.columns.name = ''

print(pvt_shsvp.shape)
display(pvt_shsvp.head())
```

### GASTPE table

```python
esc_ids = pvt_esc['school_id'].unique().tolist()
shs_ids = pvt_shsvp['school_id'].unique().tolist()

# SHSVP delivering schools that are ESC delivering as well
ids_xing = set(shs_ids).intersection(set(esc_ids))
print(len(ids_xing))

# tmp_esc = pvt_esc.copy()
# tmp_shsvp = pvt_shsvp.copy()

concat_pvt = pvt_shsvp.merge(
    pvt_esc,
    left_on='school_id',
    right_on='school_id',
    how='outer'
)
print(concat_pvt.shape)

mask_esc = concat_pvt['school_id'].isin(pvt_shsvp['school_id'].values.tolist())
concat_pvt.loc[mask_esc, 'shsvp_delivering'] = True
concat_pvt.loc[~mask_esc, 'shsvp_delivering'] = False

mask_esc = concat_pvt['school_id'].isin(pvt_esc['school_id'].values.tolist())
concat_pvt.loc[mask_esc, 'esc_delivering'] = True
concat_pvt.loc[~mask_esc, 'esc_delivering'] = False

def gastpe_school_name(row):
    name_esc = row['school_name_esc']
    name_shsvp = row['school_name_shsvp']

    if name_esc and str(name_shsvp) == 'nan':
        return name_esc
    elif str(name_esc) == 'nan' and  name_shsvp:
        return name_shsvp
    elif name_esc and name_shsvp:
        return name_esc

concat_pvt['school_name_gastpe'] = concat_pvt.apply(gastpe_school_name, axis=1)
```

### Final validation columns

```python
tmp_val = mrg_furn.copy()

# We already completed the validation column for coordinates
mask_id = tmp_val['school_id'].str.contains(r'^\d{6}$')
tmp_val.loc[mask_id, 'school_id_valid'] = True
tmp_val.loc[~mask_id, 'school_id_valid'] = False
```

```python
# Then we generate a final validation columns using our validation columns
# including coordinates_valid
cols_validation = [col for col in tmp_val.columns if re.search(r'has|valid', col, flags=re.IGNORECASE)]
mask_all_valid = tmp_val[cols_validation].apply(lambda row: all(row), axis=1)
tmp_val.loc[mask_all_valid, 'all_valid'] = True
tmp_val.loc[~mask_all_valid, 'all_valid'] = False
print(cols_validation)
print(tmp_val.shape)
```
