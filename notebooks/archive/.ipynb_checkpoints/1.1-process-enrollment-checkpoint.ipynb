```python
pattern = r'enrollment.csv$'
dirpath = 'data/processed'
filenames = os.listdir(dirpath)
filepaths = [os.path.join(dirpath, fname) for fname in filenames if re.search(pattern, fname, re.IGNORECASE)]

# List of dataframes from the CSVs
dfs_enr = [
    pd.read_csv(fpath) for fpath in filepaths
]
print(len(dfs_enr))
```

# 2.0. Enrollment Count

## 2.1. Validation

```python
# Let's inspect the shape of the dataframe of every CSV. We're particularly checking if
# the three CSVs have the same number of columns
dfs_columns = []
sys = []
for fpath, df in zip(filepaths, dfs_enr):
    pattern_sy = r'sy_(.*?)_school'
    sy = re.findall(pattern_sy, fpath, re.IGNORECASE)[0]
    print(f"Shape of {sy} dataframe: {df.shape}")

    df_columns = set(df.columns.tolist())
    dfs_columns.append(df_columns)
    sys.append(sy)

# Without looking at every file's columns, let's also make sure they all share the same column names
sy1, sy2, sy3 = dfs_columns
uniq_to_each = (sy1 - sy2 - sy3, sy2 - sy1 - sy3, sy3 - sy1 - sy2)
all_unique = (sy1 | sy2 | sy3) - (sy1 & sy2 & sy3)
print(uniq_to_each)
print(all_unique)
```

## 2.2. Consolidation

```python
# Let's view all the columns of one dataframe
display(dfs_enr[0].columns)
```

```python
columns_enr = dfs_enr[0].loc[:, 'k_male':].columns.tolist()
columns_xtotal = [col for col in columns_enr if not re.search(r'_total$', col)]
print(len(columns_xtotal))
print(columns_xtotal)
```

```python
elem = ['k'] + ['g'+str(n) for n in range(1,7)]
jhs = ['g'+str(n) for n in range(7,11)]
shs= ['g11', 'g12']

educ_lvl_enc = {
    'elementary': elem,
    'junior_high_school': jhs,
    'senior_high_school': shs
}

categorical_enc = {
    'grade_level': elem+jhs+shs,
    'education_level': list(educ_lvl_enc.keys()),
    'school_year': sys
}

def define_educ_level(dict_enc, x):
    """Defines the education level based on grade level of enrollment."""
    for k, v in dict_enc.items():
        if x in v:
            return k

dict_dfs_enr = {}
for fpath, sy, df in zip(filepaths, sys, dfs_enr):
    df_sch_ids = df['school_id'].values

    df_enr_sy = []
    for i, lvl in zip(range(0, 26, 2), elem+jhs+shs):
        col_m, col_f = columns_enr[i], columns_enr[i+1]
        # print(col_m, col_f)
        enr_m, enr_f = df[col_m], df[col_f]
        enr_lvl = df[col_m] + df[col_f]
        enr_lvl.name = lvl
        
        df_enr_sy.append(enr_lvl)
        # display(enr_lvl)
    df_enr_sy = pd.concat(df_enr_sy, axis=1)
    val_vars = df_enr_sy.columns
    df_enr_sy.index = df_sch_ids
    df_enr_sy.index.name = 'school_id'
    
    # Wide to long
    df_pvt_enr = df_enr_sy.reset_index().melt(
        id_vars='school_id', value_vars=val_vars,
        var_name='grade_level', value_name='enrollment_count'
    )

    # Drop rows with 0 enrollment values
    mask = df_pvt_enr['enrollment_count'] > 1
    df_pvt_enr = df_pvt_enr.loc[mask].reset_index(drop=True)

    # Add as new column the education level per grade level
    df_pvt_enr['education_level'] = df_pvt_enr['grade_level'].apply(lambda x: define_educ_level(educ_lvl_enc, x))

    # Add as new column the school year of the enrollment values
    df_pvt_enr['school_year'] = sy

    # Define data types
    df_pvt_enr['enrollment_count'] = df_pvt_enr['enrollment_count'].astype(int)

    for col in df_pvt_enr.columns:
        if col != 'enrollment_count':
            df_pvt_enr[col] = df_pvt_enr[col].astype(str)

    # Further define categorical columns
    columns_cat = ['grade_level', 'education_level', 'school_year']
    for col in df_pvt_enr.columns:
        if col in columns_cat:
            df_pvt_enr[col] = df_pvt_enr[col].astype('category')
            df_pvt_enr[col] = df_pvt_enr[col].cat.set_categories(
                categorical_enc[col],
                ordered=True
            )
    
    print(df_pvt_enr.shape)
    # display(df_pvt_enr.head())

    dict_dfs_enr[sy] = df_pvt_enr
    # break
```

# 3.0. School Information

## 3.1. Consolidation

```python
columns_info = dfs_enr[0].loc[:, :'sector'].columns.tolist()
print(columns_info)
```

```python
dfs_info = []
dict_sch_ids = {}
for df, sy in zip(dfs_enr, sys):
    df_temp = df[columns_info].copy()
    df_cols = df_temp.columns

    for col in df_cols:
        df_temp[col] = df_temp[col].astype(str)

    sch_ids = df_temp['school_id'].values.tolist()
    dict_sch_ids[sy] = sch_ids
    dfs_info.append(df_temp)

dfs_info = pd.concat(dfs_info)

# We want to build a table of information of unique school ids
dfs_info = dfs_info.drop_duplicates(subset='school_id')
print(dfs_info.shape)
display(dfs_info.head())
```

```python
print(dict_sch_ids.keys())
```

## 3.2. Validation

```python
def is_info_in_sy(school_ids, x):
    if x in school_ids:
        return True
    else:
        return False

# We'd like to tag schools whose information is or is not present in specific school years
for sy in sys:
    col_name = 'in_enr_' + sy
    print(col_name)
    school_ids = dict_sch_ids.get(sy)
    dfs_info[col_name] = dfs_info['school_id'].apply(
        lambda x: is_info_in_sy(school_ids, x)
    )
```

```python
def is_present_in_all_sy(x):
    if all(x):
        return True
    else:
        return False

# Show instances where a school is NOT present in all school years
mask = dfs_info[['in_2022_2023','in_2023_2024','in_2024_2025']].apply(is_present_in_all_sy, axis=1)
dfs_not = dfs_info.loc[~mask]
print(dfs_not.shape)
display(dfs_not.sample(10))
```

## 3.3. Output

```python
# Save in a format that will retain data types
fname = 'enrollment_information_3yrs.parquet'
outpath = os.path.join(OUTPUT_DIR, fname)

dfs_info.to_parquet(outpath, partition_cols='old_region')
```
