```python
import pandas as pd
import os

# Inspect SY 2023-2024 Gr 7 Enrollees
fpath = '../data/public/SY 2023-2024 Gr 7 Enrollees.xlsx'
print(f'Inspecting: {fpath}')
print('='*80)

# Read the file
df = pd.read_excel(fpath, sheet_name=0)

print(f'\nShape: {df.shape} (rows, columns)')
print(f'\nColumns ({len(df.columns)}):')
for i, col in enumerate(df.columns, 1):
    print(f'  {i:2d}. {col}')

print(f'\n\nFirst 5 rows:')
display(df.head(5))

print(f'\n\nData types:')
print(df.dtypes)

print(f'\n\nBasic statistics:')
print(df.describe())

print(f'\n\nNon-null counts:')
print(df.count())
```
