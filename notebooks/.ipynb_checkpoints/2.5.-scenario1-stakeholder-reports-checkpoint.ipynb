## 2.0 - Generate Reports

### 2.1 - Report 1: Public JHS Congestion Summary

One row per public JHS showing baseline congestion, intervention results, and top 3 private partners.

```python
# Aggregate redirections by destination (public JHS)
greedy_by_dest = redirections_greedy.groupby('school_id_destination').agg(
    redirected_greedy=('count_redirected', 'sum')
).reset_index()

lp_by_dest = redirections_lp.groupby('school_id_destination').agg(
    redirected_lp=('count_redirected_lp', 'sum')
).reset_index()

# Build Report 1 base
report1 = df_cong[['school_id', 'school_name', 'division', 'old_region', 
                   'enrollment_jhs', 'seat_count', 'count_aisle_learner_jhs']].copy()
report1.columns = ['school_id', 'school_name', 'division', 'region',
                   'enrollment_jhs', 'seat_count', 'aisle_learners_baseline']

# Merge redirections
report1 = report1.merge(greedy_by_dest, left_on='school_id', right_on='school_id_destination', how='left')
report1 = report1.merge(lp_by_dest, left_on='school_id', right_on='school_id_destination', how='left')
report1.drop(columns=['school_id_destination_x', 'school_id_destination_y'], inplace=True, errors='ignore')

# Fill NaN with 0
report1['redirected_greedy'] = report1['redirected_greedy'].fillna(0).astype(int)
report1['redirected_lp'] = report1['redirected_lp'].fillna(0).astype(int)

# Calculate after-optimization metrics
report1['aisle_learners_greedy'] = (report1['aisle_learners_baseline'] - report1['redirected_greedy']).clip(lower=0)
report1['aisle_learners_lp'] = (report1['aisle_learners_baseline'] - report1['redirected_lp']).clip(lower=0)

report1['pct_reduction_greedy'] = np.where(
    report1['aisle_learners_baseline'] > 0,
    (report1['redirected_greedy'] / report1['aisle_learners_baseline'] * 100).round(1),
    0
)
report1['pct_reduction_lp'] = np.where(
    report1['aisle_learners_baseline'] > 0,
    (report1['redirected_lp'] / report1['aisle_learners_baseline'] * 100).round(1),
    0
)

report1['is_fully_relieved'] = report1['aisle_learners_lp'] == 0

print(f"Report 1 base shape: {report1.shape}")
```

```python
# Add top 3 partners from Greedy algorithm
# Get private school info for names and tuition
private_info = sch_info[sch_info['sector'] == 'Private'][['school_id', 'school_name']].copy()
private_info.columns = ['private_school_id', 'private_school_name']

# Get tuition from redirection_options
if 'tuition_jhs' in redirection_options.columns:
    tuition_lookup = redirection_options.groupby('private_esc_id')['tuition_jhs'].first().to_dict()
else:
    tuition_lookup = {}

# Aggregate Greedy redirections by (destination, private ESC)
greedy_pairs = redirections_greedy.groupby(['school_id_destination', 'private_esc_id']).agg(
    count_redirected=('count_redirected', 'sum'),
    distance_km=('distance_to_private_km', 'mean')
).reset_index()

# Rank partners by count_redirected for each destination
greedy_pairs['rank'] = greedy_pairs.groupby('school_id_destination')['count_redirected'].rank(
    method='first', ascending=False
).astype(int)

# Filter to top 3
top3 = greedy_pairs[greedy_pairs['rank'] <= 3].copy()

# Add private school names
top3 = top3.merge(private_info, left_on='private_esc_id', right_on='private_school_id', how='left')
top3['tuition'] = top3['private_esc_id'].map(tuition_lookup)

# Pivot to wide format for top 3 partners
for i in [1, 2, 3]:
    partner_i = top3[top3['rank'] == i][['school_id_destination', 'private_esc_id', 
                                          'private_school_name', 'count_redirected', 
                                          'distance_km', 'tuition']].copy()
    partner_i.columns = ['school_id', f'top_partner_{i}_id', f'top_partner_{i}_name',
                         f'top_partner_{i}_redirected', f'top_partner_{i}_distance_km',
                         f'top_partner_{i}_tuition']
    report1 = report1.merge(partner_i, on='school_id', how='left')

# Sort by aisle_learners_baseline descending
report1 = report1.sort_values('aisle_learners_baseline', ascending=False)

print(f"Report 1 final shape: {report1.shape}")
report1.head()
```

### 2.2 - Report 2: Public-Private Redirection Pairs

Detailed view of all (public JHS, private ESC) partnerships from LP optimization.

```python
# Get school info for names and divisions
public_info = sch_info[sch_info['sector'] == 'Public'][['school_id', 'school_name', 'division']].copy()
public_info.columns = ['public_jhs_id', 'public_jhs_name', 'public_jhs_division']

private_info_full = sch_info[sch_info['sector'] == 'Private'][['school_id', 'school_name', 'division']].copy()
private_info_full.columns = ['private_esc_id', 'private_esc_name', 'private_esc_division']

# Aggregate LP redirections by (destination, private ESC)
report2_base = redirections_lp.groupby(['school_id_destination', 'private_esc_id']).agg(
    count_redirected=('count_redirected_lp', 'sum'),
    avg_distance_km=('distance_to_private_km', 'mean'),
    num_origin_schools=('school_id_origin', 'nunique')
).reset_index()

# Add public JHS info
report2 = report2_base.merge(
    public_info, left_on='school_id_destination', right_on='public_jhs_id', how='left'
)

# Add aisle learners from congestion table
aisle_lookup = df_cong.set_index('school_id')['count_aisle_learner_jhs'].to_dict()
report2['public_jhs_aisle_learners'] = report2['school_id_destination'].map(aisle_lookup)

# Add private ESC info
report2 = report2.merge(
    private_info_full, on='private_esc_id', how='left'
)

# Add slot info from esc_available
slot_info = esc_available[['school_id', 'total_count_slots', 'available_slots']].copy()
slot_info.columns = ['private_esc_id', 'total_count_slots', 'available_slots_before']
report2 = report2.merge(slot_info, on='private_esc_id', how='left')

# Calculate percentage of public school relief from this private partner
total_relief_by_public = report2.groupby('school_id_destination')['count_redirected'].transform('sum')
report2['pct_of_public_relief'] = (report2['count_redirected'] / total_relief_by_public * 100).round(1)

# Select and order columns
report2 = report2[[
    'public_jhs_id', 'public_jhs_name', 'public_jhs_division', 'public_jhs_aisle_learners',
    'private_esc_id', 'private_esc_name', 'private_esc_division',
    'total_count_slots', 'available_slots_before', 'count_redirected',
    'avg_distance_km', 'pct_of_public_relief', 'num_origin_schools'
]].copy()

report2 = report2.sort_values('count_redirected', ascending=False)

print(f"Report 2 shape: {report2.shape}")
report2.head()
```

### 2.3 - Report 3A: Excluded Public Schools

Document which public JHS were NOT included in optimization.

```python
# All public JHS in NCR/Region IV-A
all_public_jhs = sch_info[
    (sch_info['sector'] == 'Public') & 
    (sch_info['offers_jhs'] == True) &
    (sch_info['old_region'].isin(['NCR', 'Region IV-A']))
][['school_id', 'school_name', 'division', 'old_region']].copy()

# Schools in congestion analysis
included_schools = set(df_cong['school_id'])

# Find excluded schools
excluded_public = all_public_jhs[~all_public_jhs['school_id'].isin(included_schools)].copy()

# Determine exclusion reason
# Check if school is not congested (has seats >= enrollment)
congested_ids = set(df_cong[df_cong['count_aisle_learner_jhs'] > 0]['school_id'])
non_congested_ids = set(df_cong[df_cong['count_aisle_learner_jhs'] == 0]['school_id'])

def get_exclusion_reason(school_id):
    if school_id in non_congested_ids:
        return 'Not congested (no aisle learners)'
    elif school_id not in included_schools:
        return 'Missing enrollment or coordinate data'
    else:
        return 'Unknown'

excluded_public['exclusion_reason'] = excluded_public['school_id'].apply(get_exclusion_reason)
excluded_public.columns = ['school_id', 'school_name', 'division', 'region', 'exclusion_reason']

# Also add non-congested schools from df_cong as "excluded" from optimization
non_congested = df_cong[df_cong['count_aisle_learner_jhs'] == 0][['school_id', 'school_name', 'division', 'old_region']].copy()
non_congested['exclusion_reason'] = 'Not congested (no aisle learners)'
non_congested.columns = ['school_id', 'school_name', 'division', 'region', 'exclusion_reason']

report3a = pd.concat([excluded_public, non_congested], ignore_index=True).drop_duplicates(subset=['school_id'])

print(f"Report 3A shape: {report3a.shape}")
report3a['exclusion_reason'].value_counts()
```

### 2.4 - Report 3B: Excluded Private ESC Schools

Document which private ESC schools were NOT used in optimization.

```python
# All ESC schools with available slots
all_esc = esc_available[['school_id', 'total_count_slots', 'current_beneficiaries', 'available_slots']].copy()

# Add school info
all_esc = all_esc.merge(
    sch_info[['school_id', 'school_name', 'division', 'old_region']],
    on='school_id', how='left'
)

# Schools used in LP optimization
used_schools = set(redirections_lp['private_esc_id'])

# Find excluded
excluded_esc = all_esc[~all_esc['school_id'].isin(used_schools)].copy()

# Determine exclusion reason
def get_esc_exclusion_reason(row):
    if row['available_slots'] <= 0:
        return 'No available slots (fully occupied)'
    elif row['old_region'] not in ['NCR', 'Region IV-A']:
        return 'Outside NCR/Region IV-A'
    else:
        return 'Not selected by LP (no nearby congested schools or slots used elsewhere)'

excluded_esc['exclusion_reason'] = excluded_esc.apply(get_esc_exclusion_reason, axis=1)

report3b = excluded_esc[['school_id', 'school_name', 'division', 'old_region',
                         'total_count_slots', 'current_beneficiaries', 'available_slots',
                         'exclusion_reason']].copy()
report3b.columns = ['school_id', 'school_name', 'division', 'region',
                    'total_slots', 'current_beneficiaries', 'available_slots',
                    'exclusion_reason']

print(f"Report 3B shape: {report3b.shape}")
report3b['exclusion_reason'].value_counts()
```

### 2.5 - Report 4A: Student Flow Changes (Greedy)

Show how student flows from origin ES to destination public JHS change after Greedy optimization.

```python
flow_to_congested.head()
```

```python
# Start from flow_to_congested (flows to congested public JHS)
# Calculate total from non-beneficiary + esc_beneficiary columns
report4a_base = flow_to_congested[['school_id_origin', 'school_id_destination', 
                                    'count_non_beneficiary', 'count_esc_beneficiary']].copy()
report4a_base['original_total'] = report4a_base[['count_non_beneficiary','count_esc_beneficiary']].sum(axis=1)  # + report4a_base['count_esc_beneficiary']
report4a_base['original_non_beneficiaries'] = report4a_base['count_non_beneficiary']

# Aggregate greedy redirections by (origin, destination)
greedy_by_flow = redirections_greedy.groupby(['school_id_origin', 'school_id_destination']).agg(
    students_redirected=('count_redirected', 'sum'),
    num_private_partners=('private_esc_id', 'nunique')
).reset_index()

# Get top private ESC for each flow
top_private_greedy = redirections_greedy.loc[
    redirections_greedy.groupby(['school_id_origin', 'school_id_destination'])['count_redirected'].idxmax()
][['school_id_origin', 'school_id_destination', 'private_esc_id', 'count_redirected', 'distance_to_private_km']].copy()
top_private_greedy.columns = ['school_id_origin', 'school_id_destination', 
                               'top_private_esc_id', 'top_private_count', 'distance_origin_to_top_private_km']

# Merge
report4a = report4a_base.merge(greedy_by_flow, on=['school_id_origin', 'school_id_destination'], how='left')
report4a = report4a.merge(top_private_greedy, on=['school_id_origin', 'school_id_destination'], how='left')

# Fill NaN
report4a['students_redirected'] = report4a['students_redirected'].fillna(0).astype(int)
report4a['num_private_partners'] = report4a['num_private_partners'].fillna(0).astype(int)

# Calculate remaining
report4a['remaining_non_beneficiaries'] = (report4a['original_non_beneficiaries'] - report4a['students_redirected']).clip(lower=0)
report4a['pct_redirected'] = np.where(
    report4a['original_non_beneficiaries'] > 0,
    (report4a['students_redirected'] / report4a['original_non_beneficiaries'] * 100).round(1),
    0
)

# Add school names
origin_names = sch_info[['school_id', 'school_name', 'division']].copy()
origin_names.columns = ['school_id_origin', 'school_name_origin', 'division_origin']
dest_names = sch_info[['school_id', 'school_name', 'division']].copy()
dest_names.columns = ['school_id_destination', 'school_name_destination', 'division_destination']
private_names = sch_info[sch_info['sector'] == 'Private'][['school_id', 'school_name']].copy()
private_names.columns = ['top_private_esc_id', 'top_private_esc_name']

report4a = report4a.merge(origin_names, on='school_id_origin', how='left')
report4a = report4a.merge(dest_names, on='school_id_destination', how='left')
report4a = report4a.merge(private_names, on='top_private_esc_id', how='left')

# Reorder columns
report4a = report4a[[
    'school_id_origin', 'school_name_origin', 'division_origin',
    'school_id_destination', 'school_name_destination', 'division_destination',
    'original_total', 'count_esc_beneficiary', 'original_non_beneficiaries',
    'students_redirected', 'remaining_non_beneficiaries', 'pct_redirected',
    'num_private_partners', 'top_private_esc_id', 'top_private_esc_name',
    'top_private_count', 'distance_origin_to_top_private_km'
]].copy()

report4a.columns = [
    'school_id_origin', 'school_name_origin', 'division_origin',
    'school_id_destination', 'school_name_destination', 'division_destination',
    'original_total', 'original_beneficiaries', 'original_non_beneficiaries',
    'students_redirected', 'remaining_non_beneficiaries', 'pct_redirected',
    'num_private_partners', 'top_private_esc_id', 'top_private_esc_name',
    'top_private_count', 'distance_origin_to_top_private_km'
]

report4a = report4a.sort_values('students_redirected', ascending=False)

print(f"Report 4A shape: {report4a.shape}")
report4a.head()
```

### 2.6 - Report 4B: Student Flow Changes (LP)

Show how student flows change after LP optimization (same structure as 4A).

```python
# Similar to 4A but with LP redirections
report4b_base = flow_to_congested[['school_id_origin', 'school_id_destination', 
                                    'count_non_beneficiary', 'count_esc_beneficiary']].copy()
report4b_base['original_total'] = report4b_base[['count_non_beneficiary','count_esc_beneficiary']].sum(axis=1)
report4b_base['original_non_beneficiaries'] = report4b_base['count_non_beneficiary']

# Aggregate LP redirections by (origin, destination)
lp_by_flow = redirections_lp.groupby(['school_id_origin', 'school_id_destination']).agg(
    students_redirected=('count_redirected_lp', 'sum'),
    num_private_partners=('private_esc_id', 'nunique')
).reset_index()

# Get top private ESC for each flow
top_private_lp = redirections_lp.loc[
    redirections_lp.groupby(['school_id_origin', 'school_id_destination'])['count_redirected_lp'].idxmax()
][['school_id_origin', 'school_id_destination', 'private_esc_id', 'count_redirected_lp', 'distance_to_private_km']].copy()
top_private_lp.columns = ['school_id_origin', 'school_id_destination', 
                           'top_private_esc_id', 'top_private_count', 'distance_origin_to_top_private_km']

# Merge
report4b = report4b_base.merge(lp_by_flow, on=['school_id_origin', 'school_id_destination'], how='left')
report4b = report4b.merge(top_private_lp, on=['school_id_origin', 'school_id_destination'], how='left')

# Fill NaN
report4b['students_redirected'] = report4b['students_redirected'].fillna(0).astype(int)
report4b['num_private_partners'] = report4b['num_private_partners'].fillna(0).astype(int)

# Calculate remaining
report4b['remaining_non_beneficiaries'] = (report4b['original_non_beneficiaries'] - report4b['students_redirected']).clip(lower=0)
report4b['pct_redirected'] = np.where(
    report4b['original_non_beneficiaries'] > 0,
    (report4b['students_redirected'] / report4b['original_non_beneficiaries'] * 100).round(1),
    0
)

# Add school names (reuse from 4A)
report4b = report4b.merge(origin_names, on='school_id_origin', how='left')
report4b = report4b.merge(dest_names, on='school_id_destination', how='left')
report4b = report4b.merge(private_names, on='top_private_esc_id', how='left')

# Reorder columns
report4b = report4b[[
    'school_id_origin', 'school_name_origin', 'division_origin',
    'school_id_destination', 'school_name_destination', 'division_destination',
    'original_total', 'count_esc_beneficiary', 'original_non_beneficiaries',
    'students_redirected', 'remaining_non_beneficiaries', 'pct_redirected',
    'num_private_partners', 'top_private_esc_id', 'top_private_esc_name',
    'top_private_count', 'distance_origin_to_top_private_km'
]].copy()

report4b.columns = [
    'school_id_origin', 'school_name_origin', 'division_origin',
    'school_id_destination', 'school_name_destination', 'division_destination',
    'original_total', 'original_beneficiaries', 'original_non_beneficiaries',
    'students_redirected', 'remaining_non_beneficiaries', 'pct_redirected',
    'num_private_partners', 'top_private_esc_id', 'top_private_esc_name',
    'top_private_count', 'distance_origin_to_top_private_km'
]

report4b = report4b.sort_values('students_redirected', ascending=False)

print(f"Report 4B shape: {report4b.shape}")
report4b.head()
```

### 2.7 - Report 5: Slot Expansion Priorities

Identify which private ESC schools should receive additional slots to maximize congestion reduction.

**Note:** This report uses pre-computed blocked demand analysis from notebook 2.4.

```python
# Report 5 uses pre-computed blocked demand analysis from notebook 2.4
# The analysis identifies capacity-constrained schools (>=95% utilized) and calculates
# unique_blocked_demand per school (students assigned to each school without double-counting)

report5 = blocked_demand_analysis.copy()

print(f"Report 5 shape: {report5.shape}")
print(f"Capacity-constrained schools: {len(report5):,}")
print(f"Total unique blocked demand: {report5['unique_blocked_demand'].sum():,.0f}")
print(f"\nColumns: {report5.columns.tolist()}")
```

```python
# Display top 10 schools for slot expansion
print("Top 10 schools for slot expansion:")
print("="*80)

display_cols = ['private_esc_name', 'division', 'slots_used_lp', 'unique_blocked_demand', 
                'tuition_jhs', 'avg_priority_score','private_jhs_enrollment']

# Filter to columns that exist
display_cols = [c for c in display_cols if c in report5.columns]

report5[display_cols].head(10)
```

```python
# Summary by region
print("\nSlot expansion priorities by region:")
print("-"*60)

region_summary = report5.groupby('region').agg({
    'private_esc_id': 'count',
    'unique_blocked_demand': 'sum',
    'slots_used_lp': 'sum'
}).reset_index()
region_summary.columns = ['Region', 'Schools', 'Blocked Demand', 'Slots Used by LP']
region_summary = region_summary.sort_values('Blocked Demand', ascending=False)
display(region_summary)
```

```python
# Interpretation guidance
print("\n" + "="*70)
print("INTERPRETING REPORT 5 - SLOT EXPANSION PRIORITIES")
print("="*70)

print("""
Key Columns:
- unique_blocked_demand: Students assigned EXCLUSIVELY to this school 
  (no double-counting across schools)
- slots_used_lp: How many slots were used by LP optimization
- utilization_rate: Percentage of available slots used (>=95% = constrained)
- avg_priority_score: Lower = better fit (nearer + cheaper + larger)

Interpretation:
- Schools with HIGH unique_blocked_demand should be prioritized for slot expansion
- Schools with LOW slots_used_lp but HIGH unique_blocked_demand are 
  severely under-allocated relative to demand
- avg_priority_score shows how "natural" a fit the school is for nearby students

Priority Score Formula:
  priority_score = norm_distance + TUITION_WEIGHT × norm_tuition - SIZE_WEIGHT × size_bonus
  (Lower = better: nearer + cheaper + larger schools prioritized)
""")
```
