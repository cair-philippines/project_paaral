### Impute

```python
# We will use the schools' coordinates as features for imputation
df_imp = enr_seats.copy()

df_imp = df_imp.merge(
    coordinates.drop(columns='sector'),
    on='school_id',
    how='left',
)

drop_cols = ['school_name']
df_imp.drop(columns=drop_cols, inplace=True)
df_imp.set_index('school_id', inplace=True)

print(df_imp.shape)
```

```python
display(df_imp.head(1))
```

```python
df_imp.isna().sum()
```

```python
# Uncomment to see what schools do not have seats and coordinate data
mask = df_imp['seat_count'].isna()
imputed_seats_schids = df_imp[mask].index.tolist()
print(f"Count of schools whose seats data will be imputed {len(imputed_seats_schids)}")

mask = (
    (df_imp['seat_count'].isna())
    & (df_imp['longitude'].isna())
)
ignore_imputed_coordinates = df_imp[mask].index.tolist()
display(df_imp[mask])
```

```python
# 2. Define the columns
cat_cols = ['old_region', 'division']
num_cols = ['enrollment_jhs','seat_count','longitude', 'latitude']

# 3. Build the Pipeline
# Fill missing strings, then Binary Encode
cat_transformer = Pipeline([
    ('impute_cat', SimpleImputer(strategy='constant', fill_value='missing')),
    ('binary_enc', BinaryEncoder())
])

# This creates a purely numeric matrix
preprocessor = ColumnTransformer(
    transformers=[
        ('cat_logic', cat_transformer, cat_cols),
        ('num_logic', 'passthrough', num_cols)
    ]
)

# The Final Master Pipeline
# Logic: Encode -> Scale -> Impute
final_pipeline = Pipeline([
    ('prep', preprocessor),
    ('scale', MinMaxScaler()),  # Vital so price doesn't dominate zip_code
    ('knn_impute', KNNImputer(n_neighbors=5))
])

# 4. Run the Pipeline
# This will return a NumPy array where categories are binary and NaNs are filled
processed_data = final_pipeline.fit_transform(df_imp)

# Helper to see the column names (BinaryEncoder creates multiple columns)
# Note: In newer sklearn versions, get_feature_names_out() is preferred
feat_names = final_pipeline.named_steps['prep'].get_feature_names_out()
df_final = pd.DataFrame(processed_data, columns=feat_names)

# Get unscaled imputed seats count

# 1. Get the Scaler from the pipeline
scaler = final_pipeline.named_steps['scale']

# 2. Perform the Inverse Transform
# This brings everything (binary bits and floats) back to their pre-scaled state
unscaled_data = scaler.inverse_transform(processed_data)

# 3. Convert to DataFrame to make slicing easier
feat_names = final_pipeline.named_steps['prep'].get_feature_names_out()
df_unscaled = pd.DataFrame(unscaled_data, columns=feat_names)

# 4. Extract just the continuous columns you care about
# The columns will have names like 'num_logic__price' and 'num_logic__rating'
final_seats = df_unscaled['num_logic__seat_count']

print("Imputed values at original scale:")
display(final_seats.head())
```

```python
df_final.isna().sum()
```

```python
df_imputed = df_imp.copy()
df_imputed['seat_count'] = final_seats.values

df_imputed.reset_index(inplace=True)
df_imputed.drop(columns=['longitude','latitude'], inplace=True)

# Label school IDs whose seats are imputed for posterity
df_imputed['is_seats_imputed'] = None
mask = df_imputed['school_id'].isin(imputed_seats_schids)
df_imputed.loc[mask, 'is_seats_imputed'] = True
df_imputed.loc[~mask, 'is_seats_imputed'] = False

print(df_imputed.shape)
display(df_imputed.isna().sum())
```

```python
df_imputed['is_seats_imputed'].value_counts()
```

### Congestion
Notes:
- I have been using Grade 7 enrollment, which I should **NOT** use when computing for congestion or aisle_learners. Use a schools entire JHS enrollment figures from Grade 7 to 10.

```python
df_cong = df_imputed.copy()

# Seat utilization
df_cong['seat_utilization_jhs'] = df_cong['enrollment_jhs'] / df_cong['seat_count']

# Aisle learner count
df_cong['count_aisle_learner_jhs'] = df_cong['enrollment_jhs'] - df_cong['seat_count']

# No negative aisle learner
mask = df_cong['count_aisle_learner_jhs'] <= 0
df_cong.loc[mask, 'count_aisle_learner_jhs'] = 0

# Append school name
df_cong = df_cong.merge(
    sch_info_slim[['school_id','school_name']],
    on='school_id',
    how='left'
)

print(df_cong.shape)
```
